```python
import os
from datetime import datetime as dt, timedelta as td
import requests
import pandas as pd
import openalgo
from openalgo import api as openalgo_api
from dotenv import load_dotenv

load_dotenv()

openalgo_client = openalgo_api(api_key=os.getenv("API_KEY"), host='http://127.0.0.1:5000')
```

```python
print(f"{openalgo.__version__=}")
```

```python
# Long Order
response = openalgo_client.placeorder(
    strategy="Python",
    symbol="RELIANCE",
    action="BUY",
    exchange="NSE",
    price_type="LIMIT",
    product="CNC",
    quantity="1",
    price="1384.00",
    trigger_price="1384.00",
    disclosed_quantity ="0",
)
print(response)

```
